```python
def f(x):
      return 2 * x + 1

print(f(3))
print(f(0))
print(f(-2))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 50)
y = x ** 2

plt.plot(x, y)
plt.title("y = x squared")
plt.grid(True)
plt.show()
```
