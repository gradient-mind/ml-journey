```python
nums=[4,9,1,7]
print(len(nums))
print(nums[0])
print(nums[len(nums)-1])
for m in nums:
  sq=m**2
  print(sq,"is square of",m)
total=0
for m in nums:
  total=total+m
print("sum of the numbers is",total)
max=0
for m in nums:
   if m > max:
     max=m
   else:
    max=max
print("largest value is",max)
count_even=0
for m in nums:
  if m%2==0:
   count_even=count_even+1
  else:
   count_even=count_even
print("number of even numbers is",count_even)


```
